```python
needed_columns = [
    "Origin",
    "FlightDate",
    "Reporting_Airline",
    "Flight_Number_Reporting_Airline",
    "CRSDepTime",
    "DepTime",
    "DepDelay",
    "DepDel15",
    "Cancelled",
    "Diverted",
    "CarrierDelay",
    "WeatherDelay",
    "NASDelay",
    "SecurityDelay",
    "LateAircraftDelay"
]

[col for col in needed_columns if col not in df.columns]
```

```python
jfk = df[df["Origin"] == "JFK"].copy()

jfk.shape
```

```python
jfk.head()
```

```python
len(jfk)
```

```python
completed_jfk = jfk[jfk["Cancelled"] == 0]

len(completed_jfk)
```

```python
delayed_jfk = completed_jfk[completed_jfk["DepDel15"] == 1]

len(delayed_jfk)
```

```python
delay_share = len(delayed_jfk) / len(completed_jfk) * 100

delay_share
```

```python
daily_stats = (
    jfk[jfk["Cancelled"] == 0]
    .groupby("FlightDate")
    .agg(
        flights=("FlightDate", "size"),
        delayed_flights=("DepDel15", "sum")
    )
    .reset_index()
)

daily_stats["delay_share"] = (
    daily_stats["delayed_flights"] / daily_stats["flights"] * 100
)

daily_stats.head()
```
